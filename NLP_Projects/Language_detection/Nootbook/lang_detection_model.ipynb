```python
import pandas as pd
import numpy as np

df= pd.read_csv(r"/kaggle/input/datasets/basilb2s/language-detection/Language Detection.csv")
```

```python
df.shape
```

```python
df.columns
```

```python
def clean_text(text: str):
    text = text.lower()
    text =text.strip()
    return text
    
    
```

```python
df['cleaned_text']=df['Text'].apply(clean_text)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer, CountVectorizer

cvectorize= CountVectorizer(analyzer='char',ngram_range=(2,5),max_features=3000,min_df=1)
X=cvectorize.fit_transform(df['cleaned_text'])
```

```python
tfvector=TfidfVectorizer(analyzer='char',ngram_range=(2,5),max_features=3000,min_df=1)
X=tfvector.fit_transform(df['cleaned_text'])
```

```python
Y=df['Language']
```

```python
from sklearn.model_selection import train_test_split
X_Train,X_test,y_Train,y_Test=train_test_split(X,Y,test_size=0.15,random_state=42,stratify=Y)
```

```python
from sklearn.linear_model import LogisticRegression
Lr_model=LogisticRegression(max_iter=5000,C=0.5,class_weight="balanced",random_state=42)
```

```python
Lr_model.fit(X_Train,y_Train)

```

```python
y_pred=Lr_model.predict(X_test)
y_pred
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

Lr_model.score(X_test,y_Test)
```

```python
accuracy_score(y_Test,y_pred)
```

```python
print(confusion_matrix(y_Test,y_pred))
```

```python
print(classification_report(y_Test,y_pred))
```

```python
text="Olá, meu nome é Ravi"
cvText=cvectorize.transform([text])

print("Vectorized:",cvText.shape)
print("non-zero elements:", cvText.nnz)
y_pred=Lr_model.predict(cvText)
print(f" input text is '{text}' and predicted language is : {y_pred}")
```
